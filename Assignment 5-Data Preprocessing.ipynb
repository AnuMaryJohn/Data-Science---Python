```python
def outliers_iqr(df, column):
    print("\nColumn:",column)
    q1,q2,q3 = np.percentile(df[column],[25,50,75])
    print("q1",q1)
    print("q2",q2)
    print("q3",q3)
    IQR = q3-q1
    print("IQR is :" ,IQR)
    LL = q1-(1.5*IQR)
    UL = q3+(1.5*IQR)
    outliers=df[(df[column]>UL)|(df[column]<LL)]
    
    print("Lowerlimit:",LL)
    print("Upperlimit:",UL)
    print("No.of outliers:",len(outliers))

for column in outlier_cols:
    outliers_iqr(df,column)
```
