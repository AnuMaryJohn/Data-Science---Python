```python
Citric_acid=df['citric acid'].values
print('Citric_acid vector is:',Citric_acid)
```

```python
Alcohol=df['alcohol'].values
print('Alcohol vector is:',Alcohol)
```

## **Question 3**

```python
x=df[['density','alcohol']].values
cov_matrix=np.cov(x.T)
print(cov_matrix)
```

## **Question 4**

```python
eigenvalues,eigenvectors=np.linalg.eig(cov_matrix)
print('eigen values are:',eigenvalues)
print('eigen vectors are:',eigenvectors)
```

```python
print("most commonwine quality is:")
df['quality'].mode()
```

```python
df1=df['quality'].value_counts().reset_index(name="count")
print("Wine quality distribution:\n", df1)
```

### OR
